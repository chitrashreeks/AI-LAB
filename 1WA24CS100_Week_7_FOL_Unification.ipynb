```python
print("Chitra Shree K S")
print("1WA24CS100")
def unify(x, y, sub):
    if x == y:
        return True

    # Variables are x, y, z
    if x in ['x', 'y', 'z']:
        if x in sub:
            return unify(sub[x], y, sub)
        sub[x] = y
        return True

    if y in ['x', 'y', 'z']:
        if y in sub:
            return unify(x, sub[y], sub)
        sub[y] = x
        return True

    return False


def main():
    print("UNIFICATION IN FIRST ORDER LOGIC")

    exp1 = input("Enter first expression: ")
    exp2 = input("Enter second expression: ")

    try:
        p1 = exp1[:exp1.index("(")].strip()
        p2 = exp2[:exp2.index("(")].strip()

        a1 = exp1[exp1.index("(")+1:exp1.rindex(")")].split(",")
        a2 = exp2[exp2.index("(")+1:exp2.rindex(")")].split(",")

        a1 = [i.strip() for i in a1]
        a2 = [i.strip() for i in a2]

        if p1 != p2 or len(a1) != len(a2):
            print("Unification is not possible")
            return

        sub = {}

        for x, y in zip(a1, a2):
            if not unify(x, y, sub):
                print("Unification is not possible")
                return

        print("Unification is possible")
        print("MGU:", sub)

    except (ValueError, IndexError):
        print("Invalid expression")


main()
```
