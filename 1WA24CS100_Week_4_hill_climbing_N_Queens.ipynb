```python
print("Chitra Shree K S")
print("1WA24CS100")
def get_cost(state):
    """Calculates the total number of diagonal conflict pairs."""
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(i - j) == abs(state[i] - state[j]):
                cost += 1
    return cost

def format_state_string(state):
    """Formats a state array into the x0=val, x1=val... string style."""
    return ", ".join([f"x_{idx} = {val}" for idx, val in enumerate(state)])

def run_hill_climbing_4_queens():
    # --- Input Handling ---
    print("Enter the initial queen positions for rows 0, 1, 2, 3 separated by spaces.")

    user_input = input("Initial positions: ").strip()

    # Use default from image if input is empty
    if not user_input:
        state = [3, 1, 2, 0]
    else:
        state = [int(x) for x in user_input.split()]

    current_cost = get_cost(state)
    n = len(state)

    # Print header for Solution
    print(f"\nSolution: Neighbours of {format_state_string(state)} (cost is {current_cost}.)\n")

    while current_cost > 0:
        neighbors_list = []

        # 1. Generate all pairs of swaps in lexicographical order:
        # (0,1), (0,2), (0,3), (1,2), (1,3), (2,3)
        for i in range(n):
            for j in range(i + 1, n):
                neighbor_state = list(state)
                # Swap column positions
                neighbor_state[i], neighbor_state[j] = neighbor_state[j], neighbor_state[i]
                cost = get_cost(neighbor_state)

                # Keep track of indices swapped to implement tie-breaker later
                neighbors_list.append({
                    'pair': (i, j),
                    'state': neighbor_state,
                    'cost': cost,
                    'text': format_state_string(neighbor_state)
                })

        # 2. Determine the lowest cost achieved among neighbors
        min_cost = min(item['cost'] for item in neighbors_list)

        # 3. Find the first item that matches this lowest cost (Tie-breaker rule)
        # Because we iterated (i, j) in natural order, the first hit is the smallest subscript
        chosen_neighbor = None
        for item in neighbors_list:
            if item['cost'] == min_cost:
                chosen_neighbor = item
                break

        # 4. Print the generated list matching the image formatting
        for item in neighbors_list:
            action_text = ""
            # Label the neighbor as chosen only if it actually reduces or maintains the path logically
            if item == chosen_neighbor and min_cost < current_cost:
                action_text = " Choose this neighbour."

            print(f"  • {item['text']} cost is {item['cost']}.{action_text}")

        # 5. Move to the next state if an improvement is found
        if min_cost < current_cost:
            state = chosen_neighbor['state']
            current_cost = chosen_neighbor['cost']

            # If we haven't reached 0 yet, print the header for the next iteration loop
            if current_cost > 0:
                print(f"\nNeighbours of {format_state_string(state)} (cost is {current_cost}.)\n")
        else:
            print("\nLocal optimum reached. No better neighbor found.")
            break

# Execute the code
if __name__ == "__main__":
    run_hill_climbing_4_queens()
```
