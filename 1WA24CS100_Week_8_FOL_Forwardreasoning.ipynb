```python
print("Chitra Shree K S")
print("1WA24CS100")
person = input("Enter American person's name: ")
country = input("Enter country name: ")
missile = input("Enter missile name: ")
enemy = input("Enter enemy country: ")

# Initial Knowledge Base
KB = [
    ("American", person),
    ("Owns", country, missile),
    ("Missile", missile),
    ("Enemy", country, enemy)
]

# Rules
rules = [
    ([("Missile", "x")],
     ("Weapon", "x")),

    ([("Enemy", "x", enemy)],
     ("Hostile", "x")),

    ([("Missile", "x"), ("Owns", country, "x")],
     ("Sells", person, "x", country)),

    ([("American", "x"), ("Weapon", "y"),
      ("Sells", "x", "y", "z"), ("Hostile", "z")],
     ("Criminal", "x"))
]

# Match pattern with fact
def match(pattern, fact, sub):
    if pattern[0] != fact[0]:
        return None

    sub = sub.copy()
    variables = ["x", "y", "z"]

    for p, f in zip(pattern[1:], fact[1:]):
        if p in variables:
            if p in sub and sub[p] != f:
                return None
            sub[p] = f
        elif p != f:
            return None

    return sub


# Forward Chaining
def forward_chaining(KB, rules):

    while True:
        new = []

        for premises, conclusion in rules:
            subs = [{}]

            for premise in premises:
                updated = []

                for sub in subs:
                    for fact in KB:
                        result = match(premise, fact, sub)

                        if result is not None:
                            updated.append(result)

                subs = updated

            for sub in subs:
                fact = tuple(sub.get(x, x) for x in conclusion)

                if fact not in KB and fact not in new:
                    new.append(fact)

        if not new:
            break

        KB.extend(new)

        for fact in new:
            print("New Fact:", fact)

    return KB


# Run algorithm
print("\nInitial Facts:")
for fact in KB:
    print(fact)

print("\nForward Reasoning:")
result = forward_chaining(KB, rules)

print("\nFinal Result:")
if ("Criminal", person) in result:
    print(person, "is Criminal")
else:
    print("Goal Not Proved")
```
