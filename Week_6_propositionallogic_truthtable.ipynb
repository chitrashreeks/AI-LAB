```python

print("Chitra Shree K S")
print("1WA24CS100")

from itertools import product


def implies(a, b):
    return not a or b


def evaluate(expr, values):
    expr = expr.replace(" ", "")

    if "->" in expr:
        a, b = expr.split("->", 1)
        return implies(evaluate(a, values), evaluate(b, values))

    if "|" in expr:
        a, b = expr.split("|", 1)
        return evaluate(a, values) or evaluate(b, values)

    if "&" in expr:
        a, b = expr.split("&", 1)
        return evaluate(a, values) and evaluate(b, values)

    if expr.startswith("~"):
        return not evaluate(expr[1:], values)

    if expr.startswith("(") and expr.endswith(")"):
        return evaluate(expr[1:-1], values)

    if expr in values:
        return values[expr]

    raise ValueError("Invalid expression: " + expr)


def solve():
    print("\nKnowledge Base and Entailment")
    print("-" * 50)

    n = int(input("Enter number of propositions: "))
    symbols = input("Enter proposition names separated by spaces: ").split()

    if len(symbols) != n or len(set(symbols)) != n:
        print("Invalid proposition names.")
        return

    m = int(input("Enter number of KB sentences: "))
    kb = []

    for i in range(m):
        kb.append(input(f"Enter KB sentence {i + 1}: "))

    models = []
    all_rows = []

    # Generate all possible truth assignments
    for combination in product([True, False], repeat=n):
        values = dict(zip(symbols, combination))

        results = [evaluate(sentence, values) for sentence in kb]
        kb_result = all(results)

        all_rows.append((values, results, kb_result))

        if kb_result:
            models.append(values)

    # Display complete truth table
    print("\nFULL TRUTH TABLE OF KNOWLEDGE BASE")
    print("-" * 60)

    headers = symbols + [f"KB{i+1}" for i in range(m)] + ["KB"]
    print(" | ".join(headers))
    print("-" * 60)

    for values, results, kb_result in all_rows:
        row = [
            "T" if values[symbol] else "F"
            for symbol in symbols
        ]

        row += ["T" if result else "F" for result in results]
        row.append("T" if kb_result else "F")

        print(" | ".join(row))

    print("-" * 60)

    # Display models
    print("\nModels of KB:")

    for model in models:
        print({
            key: "T" if value else "F"
            for key, value in model.items()
        })

    if not models:
        print("KB has no models (it is inconsistent).")

    # Multiple query input
    q = int(input("\nEnter number of queries: "))

    for i in range(q):
        query = input(f"\nEnter query {i + 1}: ")

        if not models:
            print("Query is entailed by the Knowledge Base.")
            continue

        entailed = all(evaluate(query, model) for model in models)

        if entailed:
            print(f"Query {i + 1} is entailed by the Knowledge Base.")
        else:
            print(f"Query {i + 1} is NOT entailed by the Knowledge Base.")


if __name__ == "__main__":
    try:
        solve()
    except (ValueError, RecursionError):
        print("Invalid input. Check your propositions and expressions.")
```
